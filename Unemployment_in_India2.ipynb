```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)

file_name = 'DOC-20260812-WA0026.csv'

try:
    df = pd.read_csv(file_name)
    display(df.head())
except Exception as e:
    print(f"Error loading file: {e}")

region_column = df.columns[0]
date_column = df.columns[1]
unemp_column = df.columns[3]
employed_column = df.columns[4]

filtered_df = df[df[region_column].isin(['Uttarakhand', 'West Bengal'])]

plt.figure(figsize=(14, 6))
sns.lineplot(
    data=filtered_df,
    x=date_column,
    y=unemp_column,
    hue=region_column,
    marker='o'
)

plt.title('Unemployment Rate Trend: Uttarakhand vs West Bengal', fontsize=16, fontweight='bold')
plt.xlabel('Date / Time Period', fontsize=12)
plt.ylabel('Estimated Unemployment Rate (%)', fontsize=12)
plt.xticks(rotation=45)
plt.legend(title='Region')
plt.tight_layout()

plt.savefig('unemployment_trend_analysis.png', dpi=300)
plt.show()

summary_stats = filtered_df.groupby(region_column)[unemp_column].describe()
display(summary_stats)
```
